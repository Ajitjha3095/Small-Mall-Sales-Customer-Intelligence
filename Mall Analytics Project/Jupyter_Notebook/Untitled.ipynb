```python
# Extract unique customers
customers = df[["Customer_ID", "Customer_Type"]].drop_duplicates()

customers_path = r"C:\Users\Admin\Desktop\Workspace\Neo4j Project\Mall Analytics Project\Data set\customers.csv"
customers.to_csv(customers_path, index=False)

print("customers.csv created!")
print("Rows:", len(customers))
```

```python
# Extract unique shops
shops = df[["Shop_ID", "Shop_Name", "Category"]].drop_duplicates()

shops_path = r"C:\Users\Admin\Desktop\Workspace\Neo4j Project\Mall Analytics Project\Data set\shops.csv"
shops.to_csv(shops_path, index=False)

print("shops.csv created!")
print("Rows:", len(shops))
```

```python
transactions = df[
    [
        "Date",
        "Customer_ID",
        "Shop_ID",
        "Product_Name",
        "Quantity",
        "Unit_Price",
        "Discount",
        "Sales_Amount",
        "Payment_Method",
        "Rating"
    ]
].copy()

transactions_path = r"C:\Users\Admin\Desktop\Workspace\Neo4j Project\Mall Analytics Project\Data set\transactions.csv"
transactions.to_csv(transactions_path, index=False)

print("transactions.csv created!")
print("Rows:", len(transactions))
```
